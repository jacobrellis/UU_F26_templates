**Problem 1.** Consider the vectors

**Problem 2.** Consider the matrices
$$
A = \left[ \begin{array}{cc} 2 & 1 \\ 0 & 3 \end{array} \right],
\quad
B = \left[ \begin{array}{cc} 1 & 4 \\ -2 & 5 \end{array} \right],
\quad \text{and} \quad
C = \left[ \begin{array}{cc} 2 & -1 \\ 0 & 4 \end{array} \right].
$$
The **transpose** of a matrix $D$, $D^{\top}$, is the matrix with $D_{ij}$ swapped with $D_{ji}$, swapping rows with columns.

- a. Is $A-B = -B+A$? Compute both sides.
- b. Is $A \cdot B = B \cdot A$? Compute both products.
- c. Is $(A \cdot B)^{\top} = A^{\top} \cdot B^{\top}$? Compute both sides. What about $B^{\top} \cdot A^{\top} $?
- d. Is $(A \odot B) \odot C = A \odot (B \odot C)$? Compute both sides.

```python
mat_a = np.array([[2, 1], [0, 3]])
mat_b = np.array([[1, 4], [-2, 5]])
mat_c = np.array([[2, -1], [0,4]])
```

**Part a**

```python
# Part a
a_minus_b = mat_a - mat_b
minus_b_plus_a = -(mat_b) + mat_a

print(f'a minus b = {a_minus_b}, b minus a = {minus_b_plus_a}')
```

Yes, $A-B = -B+A$

**Part b**

```python
a_dot_b = np.inner(mat_a, mat_b)
b_dot_a = np.inner(mat_b, mat_a)

print(f'a dot b = \n{a_dot_b}\nb dot a = \n{b_dot_a}')
```

$A \cdot B \neq B \cdot A$

**Part c**

```python
adotb_transpose = (np.inner(mat_a, mat_b)).T
at_dot_bt = np.inner(mat_a.T, mat_b.T)
print(f'(a dot b) transpose = \n{adotb_transpose}\na transpose dot b transpose = \n{at_dot_bt}')
```

$(A \cdot B)^{\top} \neq A^{\top} \cdot B^{\top}$

**Part d**

```python
ab_c= (mat_a * mat_b) * mat_c
a_bc = mat_a * (mat_b * mat_c)
print(f'(a * b) * c = \n{ab_c}\na * (b * c) = \n{a_bc}')
```

Yes, $(A \odot B) \odot C = A \odot (B \odot C)$
